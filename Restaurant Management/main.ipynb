```python
import openpyxl

def create_excel():
    wb = openpyxl.Workbook()
    wb.create_sheet("Orders")
    wb.create_sheet("Waiters")
    wb.remove(wb['Sheet'])

    orders_sheet = wb["Orders"]
    orders_headers = ["Order ID", "Table Number", "Waiter Name", "Items", "Quantities", "Total", "Status"]
    for col, header in enumerate(orders_headers, 1):
        orders_sheet.cell(1, col).value = header

    waiters_sheet = wb["Waiters"]
    waiters_headers = ["Waiter ID", "Section", "Name", "Contact"]
    for col, header in enumerate(waiters_headers, 1):
        waiters_sheet.cell(1, col).value = header

    wb.save("Restaurant_Database.xlsx")
    
def Read_Orders_Database():
    Orders_Database = dict()
    file = openpyxl.load_workbook("Restaurant_Database.xlsx")
    Orders_Sheet = file["Orders"]
    
    for i in range(2, Orders_Sheet.max_row + 1):
        order_ID = Orders_Sheet.cell(i, 1).value
        if order_ID:
            Orders_Database[order_ID] = [
                str(Orders_Sheet.cell(i, 2).value),
                str(Orders_Sheet.cell(i, 3).value),
                str(Orders_Sheet.cell(i, 4).value),
                str(Orders_Sheet.cell(i, 5).value),
                str(Orders_Sheet.cell(i, 6).value),
                str(Orders_Sheet.cell(i, 7).value)
            ]
    return Orders_Database

def Read_Waiters_Database():
    Waiters_Database = dict()
    file = openpyxl.load_workbook("Restaurant_Database.xlsx")
    Waiters_Sheet = file["Waiters"]
    
    for i in range(2, Waiters_Sheet.max_row + 1):
        waiter_ID = Waiters_Sheet.cell(i, 1).value
        if waiter_ID:
            Waiters_Database[waiter_ID] = [
                [str(Waiters_Sheet.cell(i, 2).value),
                 str(Waiters_Sheet.cell(i, 3).value),
                 str(Waiters_Sheet.cell(i, 4).value)]
            ]
            
            for j in range(5, Waiters_Sheet.max_column + 1, 3):
                if Waiters_Sheet.cell(i, j).value:
                    order_data = [
                        str(Waiters_Sheet.cell(i, j).value),
                        str(Waiters_Sheet.cell(i, j+1).value),
                        str(Waiters_Sheet.cell(i, j+2).value)
                    ]
                    Waiters_Database[waiter_ID].append(order_data)
    return Waiters_Database

def Write_Orders_Database(Orders_Database):
    file = openpyxl.load_workbook("Restaurant_Database.xlsx")
    Orders_Sheet = file["Orders"]
    
    for i in range(2, Orders_Sheet.max_row + 1):
        for j in range(1, 8):
            Orders_Sheet.cell(i, j).value = ""
    
    current_row = 2
    for order_ID in Orders_Database:
        Orders_Sheet.cell(current_row, 1).value = order_ID
        for j in range(6):
            Orders_Sheet.cell(current_row, j+2).value = Orders_Database[order_ID][j]
        current_row += 1
    
    file.save("Restaurant_Database.xlsx")

def Write_Waiters_Database(Waiters_Database):
    file = openpyxl.load_workbook("Restaurant_Database.xlsx")
    Waiters_Sheet = file["Waiters"]
    
    for i in range(2, Waiters_Sheet.max_row + 1):
        for j in range(1, Waiters_Sheet.max_column + 1):
            Waiters_Sheet.cell(i, j).value = ""
    
    current_row = 2
    for waiter_ID in Waiters_Database:
        Waiters_Sheet.cell(current_row, 1).value = waiter_ID
        for j in range(3):
            Waiters_Sheet.cell(current_row, j+2).value = Waiters_Database[waiter_ID][0][j]
        
        current_col = 5
        for i in range(1, len(Waiters_Database[waiter_ID])):
            for j in range(3):
                Waiters_Sheet.cell(current_row, current_col+j).value = Waiters_Database[waiter_ID][i][j]
            current_col += 3
        current_row += 1
    
    file.save("Restaurant_Database.xlsx")

def OrderIndexInWaiterDatabase(order_ID):
    for i in Waiters_Database:
        for j in Waiters_Database[i]:
            if str(order_ID) == str(j[0]):
                Order_index = Waiters_Database[i].index(j)
                return Order_index, i

create_excel()

# Main program
print("********************************************************************")
print("*                                                                  *")
print("*             Welcome to Restaurant Billing System                 *")
print("*                                                                  *")
print("********************************************************************")

tries = 0
tries_flag = ""
while tries_flag != "Close the program":
    Orders_Database = Read_Orders_Database()
    Waiters_Database = Read_Waiters_Database()

    print("-----------------------------------------")
    print("|Enter 1 for Admin mode           |")
    print("|Enter 2 for User mode            |")
    print("|Enter 3 for exit                 |")
    print("-----------------------------------------")
    Admin_user_mode = input("Enter your mode : ")


    if Admin_user_mode == "1":
        print("*****************************************")
        print("|         Welcome to admin mode         |")
        print("*****************************************")
        Password = input("Please enter your password : ")
        while True:
            if Password == "1234":
                print("-----------------------------------------")
                print("|To manage orders Enter 1        |")
                print("|To manage waiters Enter 2       |")
                print("|To manage tables Enter 3        |")
                print("|To be back Enter E             |")
                print("-----------------------------------------")
                AdminOptions = input("Enter your choice : ")
                AdminOptions = AdminOptions.upper()

                if AdminOptions == "1":
                    print("-----------------------------------------")
                    print("|To add new order Enter 1       |")
                    print("|To display order Enter 2       |")
                    print("|To delete order Enter 3        |")
                    print("|To edit order Enter 4          |")
                    print("|To Back enter E                |")
                    print("-----------------------------------------")
                    Admin_choice = input("Enter your choice : ")
                    Admin_choice = Admin_choice.upper()

                    if Admin_choice == "1":
                        try:
                            order_ID = int(input("Enter order ID : "))
                            while order_ID in Orders_Database:
                                order_ID = int(input("This ID is unavailable, please try another ID : "))
                            
                            Table_Number = input("Enter table number : ")
                            Waiter_Name = input("Enter waiter name : ")
                            Items = input("Enter ordered items (comma separated) : ")
                            Quantity = input("Enter quantities (comma separated) : ")
                            Total = input("Enter total amount : ")
                            Status = "Pending"
                            Orders_Database[order_ID] = [Table_Number, Waiter_Name, Items, Quantity, Total, Status]
                            print("----------------------Order added successfully----------------------")
                        except:
                            print("Order ID should be an integer number")

                    elif Admin_choice == "2":
                        try:
                            order_ID = int(input("Enter order ID : "))
                            while order_ID not in Orders_Database:
                                order_ID = int(input("Incorrect ID, Please Enter order ID : "))
                            print("\nTable Number : ", Orders_Database[order_ID][0])
                            print("Waiter Name : ", Orders_Database[order_ID][1])
                            print("Items : ", Orders_Database[order_ID][2])
                            print("Quantities : ", Orders_Database[order_ID][3])
                            print("Total Amount : ", Orders_Database[order_ID][4])
                            print("Status : ", Orders_Database[order_ID][5])
                        except:
                            print("Order ID should be an integer number")

                    elif Admin_choice == "3":
                        try:
                            order_ID = int(input("Enter order ID : "))
                            while order_ID not in Orders_Database:
                                order_ID = int(input("Incorrect ID, Please Enter order ID : "))
                            Orders_Database.pop(order_ID)
                            print("----------------------Order deleted successfully----------------------")
                        except:
                            print("Order ID should be an integer number")

                    elif Admin_choice == "4":
                        try:
                            order_ID = int(input("Enter order ID : "))
                            while order_ID not in Orders_Database:
                                order_ID = int(input("Incorrect ID, Please Enter order ID : "))
                            print("-----------------------------------------")
                            print("|To Edit Table Number Enter 1    |")
                            print("|To Edit Items Enter 2          |")
                            print("|To Edit Quantities Enter 3     |")
                            print("|To Edit Total Enter 4         |")
                            print("|To Edit Status Enter 5        |")
                            print("|To be Back Enter E            |")
                            print("-----------------------------------------")
                            edit_choice = input("Enter your choice : ")
                            edit_choice = edit_choice.upper()
                            
                            if edit_choice == "1":
                                Orders_Database[order_ID][0] = input("Enter new table number : ")
                            elif edit_choice == "2":
                                Orders_Database[order_ID][2] = input("Enter new items : ")
                            elif edit_choice == "3":
                                Orders_Database[order_ID][3] = input("Enter new quantities : ")
                            elif edit_choice == "4":
                                Orders_Database[order_ID][4] = input("Enter new total : ")
                            elif edit_choice == "5":
                                Orders_Database[order_ID][5] = input("Enter new status : ")
                            
                            print("----------------------Order updated successfully----------------------")
                        except:
                            print("Order ID should be an integer number")

                elif AdminOptions == "2":
                    print("structure for waiter management")
                    pass

                elif AdminOptions == "3":
                    print("structure for table management")
                    pass

                elif AdminOptions == "E":
                    break

            elif Password != "1234":
                if tries < 2:
                    Password = input("Password incorrect, please try again : ")
                    tries += 1
                else:
                    print("Incorrect password, no more tries")
                    tries_flag = "Close the program"
                    break

            Write_Orders_Database(Orders_Database)
            Write_Waiters_Database(Waiters_Database)

    elif Admin_user_mode == "2":
        print("****************************************")
        print("|         Welcome to user mode         |")
        print("****************************************")
        while True:
            print("\n-----------------------------------------")
            print("|To view restaurant's menu Enter 1      |")
            print("|To view order status Enter 2          |")
            print("|To view table status Enter 3          |")
            print("|To view bill details Enter 4          |")
            print("|To be Back Enter E                    |")
            print("-----------------------------------------")
            UserOptions = input("Enter your choice : ")
            UserOptions = UserOptions.upper()
            
            if UserOptions == "1":
                print("Display menu")
                pass
            elif UserOptions == "2":
                print("Display order status")
                pass
            elif UserOptions == "3":
                print("Display table status")
                pass
            elif UserOptions == "4":
                print("Display bill details")
                pass
            elif UserOptions == "E":
                break

    elif Admin_user_mode == "3":
        break
    else:
        print("Please choice just 1, 2 or 3")
```
